```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
import pandas as pd
import numpy as np
```

```python

#鸢尾花数据集，查看特征，目标，样本量

li = load_iris()

print("获取特征值")
print(type(li.data))
print('-' * 50)
print(li.data.shape) # 150个样本，4个特征,一般看shape
li.data
```

```python
print(li.target)
```

```python
print(li.DESCR)
```

```python
li.feature_names    # 重点,特征名字
```

```python
li.target_names  # 目标名字
```

```python
# 进行数据的分割训练集合测试集
x_train, x_test, y_train, y_test = train_test_split(li.data, li.target,test_size=0.25, random_state=1)
```

```python
# # # 对测试集和训练集的特征值进行标准化,服务于knn fit
std = StandardScaler()
x_train = std.fit_transform(x_train)
x_test = std.transform((x_test))
```

```python
print(x_train.shape)
x_test.shape
```

# KNN

```python
knn = KNeighborsClassifier(n_neighbors=5)

knn.fit(x_train, y_train)

y_predict = knn.predict(x_test)

print('预测准确率：', knn.score(x_test, y_test))
print(y_predict, y_test)
```
